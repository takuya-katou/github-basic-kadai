```python
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= 20:
            print(f"{self.name}は{self.age}歳で、大人である。")
        else:
            print(f"{self.name}は{self.age}歳で、大人ではない。")

human_list = [Human("ichiro", 82), Human("jiro", 18), Human("saburo", 26)]

for human in human_list:
    human.check_adult()
```
